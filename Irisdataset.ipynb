```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy as stats
```

```python
# 1. display first five rows of the dataset

df = pd.read_csv('Iris.csv')
print(df.head())
print("\n")
```

```python
# 2. get random sample of  5 rows from the dataset

sample = df.sample(n=5)
print(sample)
print("\n")
```

```python
# 3. display the number of rows and columns in the dataset
rows,columns = df.shape
print(f"Rows:{rows}, Columns:{columns}")
print("\n")
```

```python
# 4. print all column names in the dataset

print(df.columns.tolist())
print("\n")
```

```python
# 5. print the dataypes of all columns in the dataset

print(df.dtypes)
print("\n")
```

```python
# 6. how many missing values exit in each column in dataset

print(df.isnull().sum())
print("\n")
```

```python
# 7. display unique species present in the "species" column

print(df['variety'].unique())
print("\n")
```

```python
# 8. count how many times each species appears in the dataset

print(df['variety'].value_counts())
print("\n")
```

```python
# 9. compute mean, median and  standard deviation of the "sepalwidthcm" column

print(f"Mean: {df['sepal.width'].mean()}")
print(f"Median: {df['sepal.width'].median()}")
print(f"Standard Deviation: {df['sepal.width'].std()}")
print("\n")
```
