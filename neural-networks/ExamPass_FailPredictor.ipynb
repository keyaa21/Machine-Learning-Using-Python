```python
import numpy as np
```

```python
def sigmoid(x):
  return 1 / (1 + np.exp(-x))
```

```python
def deriv_sigmoid(x):
  fx = sigmoid(x)
  return fx * (1 - fx)
```

```python
def mse_loss(y_true, y_pred):
  return ((y_true - y_pred)**2).mean()
```

```python
class PassFailNetwork:
  def __init__(self):
    # Scale initial random weights slightly so they don't saturate the sigmoid
    self.w1 = np.random.normal() * 0.5
    self.w2 = np.random.normal() * 0.5
    self.w3 = np.random.normal() * 0.5
    self.w4 = np.random.normal() * 0.5
    self.w5 = np.random.normal() * 0.5
    self.w6 = np.random.normal() * 0.5

    # Biases
    self.b1 = np.random.normal() * 0.5
    self.b2 = np.random.normal() * 0.5
    self.b3 = np.random.normal() * 0.5

  def feedforward(self, x):
    h1 = sigmoid(self.w1 * x[0] + self.w2 * x[1] + self.b1)
    h2 = sigmoid(self.w3 * x[0] + self.w4 * x[1] + self.b2)
    o1 = sigmoid(self.w5 * h1 + self.w6 * h2 + self.b3)
    return o1

  def train(self, data, all_y_trues):
    learning_rate = 0.3  # Stronger step size to bypass plateaus
    epochs = 1500

    for epoch in range(epochs):
      for x, y_true in zip(data, all_y_trues):
        sum_h1 = self.w1 * x[0] + self.w2 * x[1] + self.b1
        h1 = sigmoid(sum_h1)

        sum_h2 = self.w3 * x[0] + self.w4 * x[1] + self.b2
        h2 = sigmoid(sum_h2)

        sum_o1 = self.w5 * h1 + self.w6 * h2 + self.b3
        o1 = sigmoid(sum_o1)
        y_pred = o1

        # --- Calculate partial derivatives
        d_L_d_y_pred = -2 * (y_true - y_pred)

        # Neuron o1
        d_y_pred_d_w5 = h1 * deriv_sigmoid(sum_o1)
        d_y_pred_d_w6 = h2 * deriv_sigmoid(sum_o1)
        d_y_pred_d_b3 = deriv_sigmoid(sum_o1)

        d_y_pred_d_h1 = self.w5 * deriv_sigmoid(sum_o1)
        d_y_pred_d_h2 = self.w6 * deriv_sigmoid(sum_o1)

        # Neuron h1
        d_h1_d_w1 = x[0] * deriv_sigmoid(sum_h1)
        d_h1_d_w2 = x[1] * deriv_sigmoid(sum_h1)
        d_h1_d_b1 = deriv_sigmoid(sum_h1)

        # Neuron h2
        d_h2_d_w3 = x[0] * deriv_sigmoid(sum_h2)
        d_h2_d_w4 = x[1] * deriv_sigmoid(sum_h2)
        d_h2_d_b2 = deriv_sigmoid(sum_h2)

        # --- Update weights and biases
        self.w1 -= learning_rate * d_L_d_y_pred * d_y_pred_d_h1 * d_h1_d_w1
        self.w2 -= learning_rate * d_L_d_y_pred * d_y_pred_d_h1 * d_h1_d_w2
        self.b1 -= learning_rate * d_L_d_y_pred * d_y_pred_d_h1 * d_h1_d_b1

        self.w3 -= learning_rate * d_L_d_y_pred * d_y_pred_d_h2 * d_h2_d_w3
        self.w4 -= learning_rate * d_L_d_y_pred * d_y_pred_d_h2 * d_h2_d_w4
        self.b2 -= learning_rate * d_L_d_y_pred * d_y_pred_d_h2 * d_h2_d_b2

        self.w5 -= learning_rate * d_L_d_y_pred * d_y_pred_d_w5
        self.w6 -= learning_rate * d_L_d_y_pred * d_y_pred_d_w6
        self.b3 -= learning_rate * d_L_d_y_pred * d_y_pred_d_b3

      if epoch % 100 == 0:
        y_preds = np.apply_along_axis(self.feedforward, 1, data)
        loss = mse_loss(all_y_trues, y_preds)
        print("Epoch %d loss: %.4f" % (epoch, loss))

# Dataset (Shifted values)
data = np.array([[-3, -1], [3, 2], [2, 0], [-2, 0]])
all_y_trues = np.array([0, 1, 1, 0])

# Train
network = PassFailNetwork()
network.train(data, all_y_trues)

```

```python
# Test Emma and Frank
emma = np.array([4, 1])
frank = np.array([-4, -2])

print("\n--- Final Predictions ---")
print("Emma's prediction (Pass ~1.0):", network.feedforward(emma))
print("Frank's prediction (Fail ~0.0):", network.feedforward(frank))
```
